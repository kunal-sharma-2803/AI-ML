```python
%matplotlib inline 
# matplotlib is in the notebook, i.e. figures should be shown here only
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
plt.plot(); # generates the figure (plot)
```

```python
plt.plot([1,2,3,4]);
```

```python
x = [1,2,3,4]
y = [11,12,13,14]
plt.plot(x,y);
# stateless way of plotting
# this is also called pyplot API, which is less flexible than object-oriented API
# look for lifecycle of a plot in matplotlib 
```

```python
# 1st method 
fig = plt.figure() # creates a figure
#plt.show();
ax = fig.add_subplot() # adds some axes
plt.show()
```

```python
# 2nd method
fig = plt.figure() # creates a figure
ax = fig.add_axes([1,1,1,1]) 
ax.plot(x,y) # add some data
plt.show()
```

```python
# 3rd method (recommended)
fig, ax = plt.subplots()
ax.plot(x,y);
```
