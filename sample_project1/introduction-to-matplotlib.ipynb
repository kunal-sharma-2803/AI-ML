# matplotlib example workflow

```python
# 0. import matplotlib and get it ready for plotting in jupyter
%matplotlib inline
import matplotlib.pyplot as plt

# 1. prepare data
x = [1,2,3,4]
y = [11,22,33,11]

# 2. setup plot
fig, ax = plt.subplots()

# 3. plot data
ax.plot(x,y)

# 4. customise plot
ax.set(title="sample plot", xlabel="x-axis", ylabel="yaxis")

# 5. save & show (you save the whole figure)
fig.savefig("./matplotlib-plots/sample-plot.png")
```

# making figures with numpy arrays

```python
import numpy as np
```

```python
# create some data
x = np.linspace(0,10,100)
x
```

```python
# plot the data, by default plot creates a line plot
fig, ax = plt.subplots()
ax.plot(x,x**2);
```

```python
# use the same data to create a scatter plot
fig, ax = plt.subplots()
ax.scatter(x,np.exp(x));
```

```python
# another scatter plot
fig, ax = plt.subplots()
ax.scatter(x,np.sin(x));
```

```python
nut_butter_prices = {"almond":10,"peanut":8,"cashew":12}
fig, ax = plt.subplots()
ax.bar(nut_butter_prices.keys(),nut_butter_prices.values())
ax.set(title="nut butter prices",xlabel="Type",ylabel="Price");
```

```python
x = np.random.randint(1,10,size=(1,10))
y = np.random.randint(20,50,size=(1,10))
```

```python
fig, ax = plt.subplots()
ax.scatter(x,y);
```

```python
# make some data for histogram
x = np.random.randn(10000)
fig, ax = plt.subplots()
ax.hist(x);
```

```python
# subplot option 1
fig, ((ax1,ax2),(ax3,ax4)) = plt.subplots(nrows=2,ncols=2,figsize=(10,5))

# plot to each different axis
ax1.plot(x,x/2);
ax2.scatter(np.random.random(10),np.random.random(10));
ax3.bar(nut_butter_prices.keys(),nut_butter_prices.values());
ax4.hist(np.random.randn(1000));
```

```python
# sub plots 2
fig, ax = plt.subplots(nrows=2,ncols=2,figsize=(10,5))
ax[0,0].plot(x,x/2);
ax[0,1].scatter(np.random.random(10),np.random.random(10));
# same for other 2
```

# plotting form pandas dataframes

```python
import pandas as pd
```

```python
# make a dataframe
car_sales = pd.read_csv("car-sales.csv")
car_sales
```

```python
ts = pd.Series(np.random.randn(1000),index=pd.date_range("23/03/2026",periods=1000))
ts = ts.cumsum() # cumulative sum
ts.plot()
```

```python
car_sales
```

```python
car_sales.Price = car_sales.Price.str.replace('$','')
car_sales.Price = car_sales.Price.str.replace(',','')
car_sales.Price = car_sales.Price.str.replace('.','')
car_sales
```

```python
car_sales.Price = car_sales.Price.str[:-2]
car_sales
```

```python
car_sales["Sale Day"] = pd.date_range("24/03/2026",periods=len(car_sales))
car_sales
```

```python
car_sales.Price = pd.to_numeric(car_sales.Price)
```

```python
car_sales.Price.dtype
```

```python
car_sales["total sales"] = car_sales.Price.cumsum()
car_sales
```

```python
car_sales.plot(x="Sale Day",y="total sales");
```

```python
car_sales.plot(x="Odometer (KM)", y="Price", kind="scatter");
```

```python
# how about a bar graph 
x = np.random.rand(10,5)

df = pd.DataFrame(x,columns=['a','b','c','d','e'])
df
```
