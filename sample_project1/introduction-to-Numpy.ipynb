# Exercise

```python
np.random.seed(0)
sales_quantity = np.random.randint(20, size=(5,3))
sales_quantity
```

```python
weekly_sales = pd.DataFrame(sales_quantity, 
                            index=["Mon","Teus","Wed","Thurs","Fri"],
                            columns=["Almond","Peanut","Cashew"])
weekly_sales
```

```python
np.random.seed(1)
price = np.random.randint(15, size=(1,3))
price
```

```python
price_division = pd.DataFrame(price,index=["Ruppes"],columns=["Almond","Peanut","Cashew"])
price_division
```

```python
total_weekly_sales = np.dot(weekly_sales,price_division)
total_weekly_sales
```

```python
price_division_transpose = price_division.T
price_division_transpose
```

```python
daily_sales = weekly_sales.dot(price_division_transpose)
daily_sales
```

```python
weekly_sales["Total"] = daily_sales
weekly_sales
```
