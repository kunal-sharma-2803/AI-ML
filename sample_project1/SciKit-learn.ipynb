## What is SciKit-learn (sklearn) :- Python machine learning model

#### Why sklearn :-
1. built on numpy and matplotlib
2. has many in-built ML models
3. Methods to evaluate your ml models 
4. very well designed API

### Workflow :-
1. get data ready 
2. pick a model (to suit your model)
3. Fit the model to the data and make a prediction (asking the model to find patterns)
4. Evaluate the model 
5. improve through experimentation
6. Save and reload your trained model
